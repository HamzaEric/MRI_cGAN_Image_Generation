```python
from pathlib import Path

import matplotlib.pyplot as plt
import medigan
import torch
import torch.optim as optim
import torchvision
from torchvision.io import read_image
from torchvision.utils import make_grid
from tqdm.notebook import tqdm
from medigan import Generators
import os
import PIL
from PIL import Image
```

```python
generators = Generators()

keywords = ["brain", "MRI"]
search_results = generators.find_matching_models_by_values(keywords)

print("Available Models:")
print(search_results)
```

```python
TARGET_MODEL_ID = "00021_CYCLEGAN_BRAIN_MRI_T1_T2"


output_path = "./medigan_synthetic_mris"
os.makedirs(output_path, exist_ok=True)

print(f"Generating images using model {TARGET_MODEL_ID}...")

# Generate 100 images
generators.generate(
    model_id=TARGET_MODEL_ID,
    num_samples=100,
    output_path=output_path,
    install_dependencies=True
)

print(f"Done! Check the '{output_path}' directory for your images.")
```
